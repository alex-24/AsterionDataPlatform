```sql
CREATE CATALOG IF NOT EXISTS asterion;
CREATE SCHEMA IF NOT EXISTS bronze;
CREATE SCHEMA IF NOT EXISTS silver;
CREATE SCHEMA IF NOT EXISTS gold;
USE CATALOG asterion
```

```sql
CREATE TABLE IF NOT EXISTS bronze.trades (
    trade_id STRING,
    portfolio_id STRING,
    instrument_id STRING,
    trade_date STRING,
    trade_type STRING, -- "BUY"/"SELL"
    trade_quantity DECIMAL(18, 4), -- >0
    trade_price DECIMAL(18, 6), -- >0
    trade_currency STRING
)
COMMENT "timestamped trades";

CREATE TABLE IF NOT EXISTS bronze.instruments (
    instrument_id STRING,
    instrument_name STRING,
    ticker STRING,
    asset_class STRING,
    currency STRING, -- "EUR"/"USD"
    exchange STRING,
    sector STRING
)
COMMENT "instruments";

CREATE TABLE IF NOT EXISTS bronze.closing_price (
    close_date DATE,
    instrument_id STRING,
    close_price DECIMAL(18, 6), -- >0
    currency STRING
)
COMMENT "closing price of an instrument per date";

CREATE TABLE IF NOT EXISTS bronze.foreign_exchange_rate (
    rate_date DATE,
    base_currency STRING,
    quote_currency STRING,
    rate DECIMAL(18, 8) -- >0
)
COMMENT "FX conversion rates per date"
```
